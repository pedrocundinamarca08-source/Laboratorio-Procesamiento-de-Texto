
# 2. Normalización de texto

La normalización de texto consiste en transformar un texto a un formato uniforme para facilitar su posterior procesamiento.

Entre las operaciones más comunes se encuentran:

- Convertir el texto a minúsculas.
- Eliminar signos de puntuación.
- Eliminar caracteres especiales.
- Eliminar espacios innecesarios.
- Eliminar números cuando no son relevantes.

La normalización ayuda a reducir diferencias innecesarias entre palabras que realmente representan el mismo término.


```python
# Texto utilizado como ejemplo

texto_original = """
La Inteligencia Artificial está CAMBIANDO el mundo!!!
En 2026, muchas empresas utilizan IA para analizar información,
automatizar procesos y mejorar la toma de decisiones.
"""

print("TEXTO ORIGINAL:")
print(texto_original)
```

```python
def normalizar_texto(texto):
    # Convertir todo el texto a minúsculas
    texto = texto.lower()

    # Eliminar números
    texto = re.sub(r'\d+', '', texto)

    # Eliminar signos de puntuación y caracteres especiales
    texto = re.sub(r'[^\w\sáéíóúüñ]', '', texto)

    # Eliminar espacios múltiples
    texto = re.sub(r'\s+', ' ', texto)

    # Eliminar espacios al inicio y al final
    texto = texto.strip()

    return texto


texto_normalizado = normalizar_texto(texto_original)

print("TEXTO NORMALIZADO:")
print(texto_normalizado)
```

### Análisis del resultado

Después de aplicar la normalización se observa que el texto fue convertido completamente a minúsculas y se eliminaron números, signos de puntuación y espacios innecesarios.

Este procedimiento facilita las siguientes etapas del procesamiento de lenguaje natural, ya que reduce las variaciones innecesarias presentes en el texto.

# 3. Stemming

El stemming es una técnica que intenta reducir las palabras a una raíz común eliminando prefijos o sufijos.

El resultado no siempre corresponde a una palabra válida del idioma, ya que el objetivo principal es agrupar palabras que poseen una raíz similar.

Por ejemplo:

- estudiar
- estudiando
- estudiante
- estudios

pueden ser reducidas a una raíz similar.

```python
# Crear stemmer para idioma español

stemmer = SnowballStemmer("spanish")

palabras = [
    "estudiar",
    "estudiando",
    "estudiante",
    "estudios",
    "computación",
    "computadores",
    "inteligencia"
]

print("PALABRA       -> STEM")
print("-------------------------")

for palabra in palabras:
    raiz = stemmer.stem(palabra)
    print(f"{palabra:15} -> {raiz}")
```

```python
texto_stemming = """
Los estudiantes están estudiando diferentes técnicas
de inteligencia artificial y procesamiento de textos.
"""

# Normalizar
texto_stemming = normalizar_texto(texto_stemming)

# Separar el texto en palabras
tokens = word_tokenize(texto_stemming, language="spanish")

# Aplicar stemming
stems = [stemmer.stem(palabra) for palabra in tokens]

print("Palabras originales:")
print(tokens)

print("\nDespués de aplicar Stemming:")
print(stems)
```

### Análisis del resultado

Se observa que varias palabras fueron reducidas a una forma de raíz.

Por ejemplo, palabras relacionadas con **estudiar** pueden convertirse en una raíz similar como **"estudi"**.

La principal ventaja del stemming es reducir el número de palabras diferentes que debe analizar un sistema.

# 4. Eliminación de Stopwords

Las stopwords son palabras que aparecen frecuentemente en un idioma pero que normalmente aportan poca información al análisis del contenido.

algunos ejemplos son:

- el
- la
- los
- de
- y
- para
- en

Eliminar estas palabras permite concentrar el análisis en términos que contienen mayor cantidad de información.

```python
# Obtener stopwords en español

stop_words = set(stopwords.words("spanish"))

print("Cantidad de stopwords disponibles:")
print(len(stop_words))

print("\nAlgunos ejemplos:")
print(list(stop_words)[:20])
```

```python
texto = """
La inteligencia artificial es una tecnología que permite
a las computadoras realizar tareas relacionadas con el
aprendizaje y el análisis de información.
"""

# Normalizar texto
texto = normalizar_texto(texto)

# Tokenizar
tokens = word_tokenize(texto, language="spanish")

# Eliminar stopwords
tokens_sin_stopwords = [
    palabra for palabra in tokens
    if palabra not in stop_words
]

print("Texto tokenizado:")
print(tokens)

print("\nTexto sin stopwords:")
print(tokens_sin_stopwords)
```

### Análisis del resultado

Al eliminar las stopwords desaparecieron palabras frecuentes como **"la"**, **"es"**, **"una"**, **"que"**, **"a"**, **"las"**, **"con"** y **"el"**.

El texto resultante conserva principalmente las palabras relacionadas directamente con el tema analizado.

Esto permite reducir el ruido y facilita procesos posteriores como clasificación, búsqueda de palabras importantes y análisis de frecuencia.

# 5. Term Frequency (TF)

Term Frequency (TF) permite medir qué tan frecuente aparece una palabra dentro de un documento.

Se puede calcular mediante:

**TF = Número de veces que aparece el término / Número total de palabras del documento**

Un valor alto indica que una palabra aparece con mayor frecuencia dentro del texto analizado.

```python
from collections import Counter

texto_tf = """
La inteligencia artificial permite analizar datos.
La inteligencia artificial también permite automatizar procesos.
Los datos son importantes para la inteligencia artificial.
"""

# Normalizar el texto
texto_tf = normalizar_texto(texto_tf)

# Tokenizar
tokens_tf = word_tokenize(texto_tf, language="spanish")

# Eliminar stopwords
tokens_tf_limpios = [
    palabra for palabra in tokens_tf
    if palabra not in stop_words
]

# Contar palabras
frecuencias = Counter(tokens_tf_limpios)

# Número total de términos
total_terminos = len(tokens_tf_limpios)

# Calcular TF
resultado_tf = []

for palabra, frecuencia in frecuencias.items():
    tf = frecuencia / total_terminos
    resultado_tf.append([palabra, frecuencia, tf])

# Crear tabla
df_tf = pd.DataFrame(
    resultado_tf,
    columns=["Palabra", "Frecuencia", "TF"]
)

# Ordenar de mayor a menor
df_tf = df_tf.sort_values(
    by="TF",
    ascending=False
).reset_index(drop=True)

df_tf
```

### Análisis del resultado

El cálculo de Term Frequency permite identificar cuáles son las palabras que aparecen con mayor frecuencia dentro del documento.

En este ejemplo, términos como **"inteligencia"** y **"artificial"** presentan valores altos de TF, lo que indica que tienen una presencia importante dentro del texto.

Sin embargo, una palabra frecuente no necesariamente es la más importante cuando se comparan varios documentos. Para solucionar este problema se utiliza el IDF.

# 6. Inverse Document Frequency (IDF)

Inverse Document Frequency (IDF) permite determinar qué tan informativa es una palabra dentro de una colección de documentos.

Las palabras que aparecen en muchos documentos reciben menor importancia, mientras que aquellas que aparecen en pocos documentos reciben un valor mayor.

El IDF es especialmente útil cuando se combina con Term Frequency, formando la técnica conocida como **TF-IDF**.

```python
documentos = [
    "La inteligencia artificial permite analizar grandes cantidades de datos.",
    "El aprendizaje automático utiliza datos para entrenar modelos.",
    "La inteligencia artificial permite automatizar procesos empresariales."
]

for i, documento in enumerate(documentos, start=1):
    print(f"Documento {i}: {documento}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Crear vectorizador
vectorizador_idf = TfidfVectorizer(
    lowercase=True,
    stop_words=list(stop_words)
)

# Procesar documentos
matriz_tfidf = vectorizador_idf.fit_transform(documentos)

# Obtener palabras
terminos = vectorizador_idf.get_feature_names_out()

# Obtener valores IDF
valores_idf = vectorizador_idf.idf_

# Crear tabla
df_idf = pd.DataFrame({
    "Término": terminos,
    "IDF": valores_idf
})

df_idf = df_idf.sort_values(
    by="IDF",
    ascending=False
).reset_index(drop=True)

df_idf
```

### Análisis del resultado

Las palabras que aparecen en varios documentos reciben un valor IDF menor porque permiten diferenciar menos entre los textos.

Por otro lado, las palabras que aparecen solamente en uno o pocos documentos reciben un valor IDF mayor, ya que pueden aportar mayor información para distinguir su contenido.

Por esta razón, IDF complementa a la frecuencia de términos y permite reducir la importancia de palabras demasiado comunes.

# 7. Part-of-Speech Tagging (POS Tagging)

Part-of-Speech Tagging consiste en identificar la función gramatical de cada palabra dentro de una oración.

Por ejemplo, permite reconocer:

- Sustantivos
- Verbos
- Adjetivos
- Adverbios
- Pronombres
- Determinantes

Esta información es útil para comprender mejor la estructura de un texto.

```python
texto_pos = """
Los estudiantes desarrollan proyectos innovadores
utilizando inteligencia artificial.
"""

doc_pos = nlp(texto_pos)

resultado_pos = []

for token in doc_pos:
    if not token.is_space and not token.is_punct:
        resultado_pos.append([
            token.text,
            token.pos_,
            spacy.explain(token.pos_)
        ])

df_pos = pd.DataFrame(
    resultado_pos,
    columns=[
        "Palabra",
        "Etiqueta POS",
        "Descripción"
    ]
)

df_pos
```

### Análisis del resultado

El POS Tagging permitió identificar automáticamente la categoría gramatical de cada palabra.

Esta información permite que los sistemas de procesamiento de lenguaje natural comprendan mejor la estructura gramatical de una oración.

## 8. Lematización

La lematización es un proceso que reduce las palabras a su forma base o lema, a diferencia del stemming, el lema siempre será una palabra válida en el idioma.

Por ejemplo:

- 'corriendo' -> 'correr'
- 'estudiantes' -> 'estudiante'
- 'mejores' -> 'bueno'

La lematización es útil cuando se necesita una representación más precisa de las palabras, especialmente para análisis de texto donde la forma base de la palabra es importante para el significado.

```python
texto_lema = """
Los estudiantes estaban estudiando diferentes técnicas
y desarrollaron varios proyectos tecnológicos.
"""

doc_lema = nlp(texto_lema)

resultado_lemas = []

for token in doc_lema:
    if not token.is_space and not token.is_punct:
        resultado_lemas.append([
            token.text,
            token.lemma_
        ])

df_lemas = pd.DataFrame(
    resultado_lemas,
    columns=["Palabra original", "Lema"]
)

df_lemas
```

### Análisis del resultado

La lematización transformó diferentes palabras a su forma base.

La lematización suele producir resultados lingüísticamente más correctos que el stemming, aunque requiere un análisis más avanzado del idioma.

# 9. Parsing o análisis sintáctico

El parsing permite analizar la estructura de una oración y las relaciones que existen entre sus palabras.

Mediante esta técnica es posible identificar elementos como el sujeto, el verbo principal, los complementos y las dependencias gramaticales.

Esto permite comprender no solamente qué palabras aparecen en el texto, sino también cómo se encuentran relacionadas.

```python
oracion_parsing = "El estudiante desarrolla un proyecto de inteligencia artificial."

doc_parsing = nlp(oracion_parsing)

resultado_parsing = []

for token in doc_parsing:
    if not token.is_space:
        resultado_parsing.append([
            token.text,
            token.dep_,
            token.head.text,
            token.pos_
        ])

df_parsing = pd.DataFrame(
    resultado_parsing,
    columns=[
        "Palabra",
        "Dependencia",
        "Palabra principal",
        "POS"
    ]
)

df_parsing
```

```python
for token in doc_parsing:
    if token.dep_ == "ROOT":
        print("Verbo o palabra principal de la oración:")
        print(token.text)
```

### Análisis del resultado

El análisis sintáctico permitió identificar las relaciones existentes entre las palabras de la oración.

En el ejemplo, el verbo **"desarrolla"** funciona como uno de los elementos centrales de la estructura de la oración, mientras que las demás palabras mantienen diferentes relaciones gramaticales con él.

Esta técnica permite analizar la organización de una oración y resulta útil en aplicaciones que requieren comprender relaciones entre sujetos, acciones y objetos.

# 10. Named Entity Recognition (NER)

Named Entity Recognition, o reconocimiento de entidades nombradas, es una técnica que permite identificar automáticamente elementos importantes dentro de un texto.

Entre las entidades que pueden detectarse se encuentran:

- Personas
- Organizaciones
- Ciudades
- Países
- Lugares
- Fechas

NER es utilizado en buscadores, sistemas de extracción de información, análisis de noticias y asistentes virtuales.

```python
texto_ner = """
Google anunció un nuevo proyecto de inteligencia artificial
en Madrid. La empresa también tiene operaciones en Colombia
y Estados Unidos.
"""

doc_ner = nlp(texto_ner)

entidades = []

for entidad in doc_ner.ents:
    entidades.append([
        entidad.text,
        entidad.label_
    ])

df_entidades = pd.DataFrame(
    entidades,
    columns=[
        "Entidad detectada",
        "Tipo"
    ]
)

df_entidades
```

# 12. Question Answering

Question Answering es una tarea de procesamiento de lenguaje natural que permite responder preguntas automáticamente utilizando información contenida en un texto.

El sistema recibe principalmente dos elementos:

1. Un contexto que contiene información.
2. Una pregunta relacionada con ese contexto.

El modelo analiza ambos elementos e identifica la respuesta más probable.

```python
from transformers import AutoTokenizer, AutoModelForQuestionAnswering
import torch

# Modelo de Question Answering en español
nombre_modelo = "mrm8488/bert-base-spanish-wwm-cased-finetuned-spa-squad2-es"

# Cargar tokenizer y modelo
tokenizer = AutoTokenizer.from_pretrained(nombre_modelo)
modelo_qa = AutoModelForQuestionAnswering.from_pretrained(nombre_modelo)

# Contexto
contexto = """
La inteligencia artificial es una rama de la informática
que desarrolla sistemas capaces de realizar tareas que normalmente
requieren inteligencia humana.

Entre sus aplicaciones se encuentran el reconocimiento de imágenes,
el procesamiento de lenguaje natural, los vehículos autónomos
y la automatización industrial.
"""

# Pregunta
pregunta = "¿Cuáles son algunas aplicaciones de la inteligencia artificial?"

# Convertir pregunta y contexto a datos que entiende BERT
entradas = tokenizer(
    pregunta,
    contexto,
    return_tensors="pt",
    truncation=True,
    max_length=512
)

# Realizar la predicción
with torch.no_grad():
    salida = modelo_qa(**entradas)

# Obtener posición inicial y final de la respuesta
inicio = torch.argmax(salida.start_logits, dim=1).item()
fin = torch.argmax(salida.end_logits, dim=1).item()

# Extraer los tokens correspondientes a la respuesta
tokens_respuesta = entradas["input_ids"][0][inicio:fin + 1]

# Convertir tokens nuevamente a texto
respuesta = tokenizer.decode(
    tokens_respuesta,
    skip_special_tokens=True
)

print("Pregunta:")
print(pregunta)

print("\nRespuesta:")
print(respuesta)
```

### Análisis del resultado

El modelo analizó la pregunta y buscó dentro del contexto el fragmento que mejor permitía responderla.

Este tipo de técnica puede utilizarse en asistentes virtuales, sistemas de consulta de documentos y herramientas de búsqueda inteligente.

# 13. Summarization

La summarization o generación automática de resúmenes busca reducir un texto conservando sus ideas principales.

Esta técnica permite procesar grandes cantidades de información y producir versiones más cortas que mantienen el contenido relevante.

Puede utilizarse para resumir:

- Noticias.
- Documentos.
- Informes.
- Artículos.
- Investigaciones.

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch

# Modelo especializado en resumen
nombre_modelo = "facebook/bart-large-cnn"

# Cargar tokenizer y modelo
tokenizer_resumen = AutoTokenizer.from_pretrained(nombre_modelo)
modelo_resumen = AutoModelForSeq2SeqLM.from_pretrained(nombre_modelo)

# Texto que queremos resumir
texto_resumen = """
Artificial intelligence is transforming many industries around the world.
Companies are using AI systems to automate repetitive tasks, analyze large
amounts of data and improve decision-making. Machine learning algorithms can
identify patterns that would be difficult for humans to detect manually.
Artificial intelligence is also being used in healthcare, transportation,
manufacturing and education. However, organizations must consider issues
such as data quality, privacy, security and responsible use when implementing
these technologies.
"""

# Convertir el texto a tokens
entradas = tokenizer_resumen(
    texto_resumen,
    return_tensors="pt",
    max_length=1024,
    truncation=True
)

# Generar el resumen
with torch.no_grad():
    resumen_ids = modelo_resumen.generate(
        entradas["input_ids"],
        attention_mask=entradas["attention_mask"],
        max_length=70,
        min_length=25,
        num_beams=4,
        early_stopping=True
    )

# Convertir los tokens nuevamente a texto
resumen = tokenizer_resumen.decode(
    resumen_ids[0],
    skip_special_tokens=True
)

print("TEXTO ORIGINAL:")
print(texto_resumen)

print("\nRESUMEN:")
print(resumen)
```

### Análisis del resultado

El modelo redujo la extensión del texto original manteniendo las ideas principales relacionadas con el uso de inteligencia artificial y sus diferentes aplicaciones.

La summarization permite obtener rápidamente una representación más corta de documentos extensos.

Sin embargo, es necesario comprobar que el resumen mantenga correctamente el significado del documento original, ya que los modelos automáticos pueden omitir información importante.

# 14. Sentence Similarity

Sentence Similarity permite calcular qué tan similares son dos oraciones según su significado.

Para realizar esta tarea, las oraciones pueden convertirse en representaciones numéricas llamadas embeddings.

Posteriormente se calcula la similitud de coseno entre estos vectores.

Un valor cercano a 1 indica una alta similitud, mientras que un valor más cercano a 0 indica una menor relación semántica.

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# Modelo multilingüe
modelo_similitud = SentenceTransformer(
    "paraphrase-multilingual-MiniLM-L12-v2"
)

oracion1 = "La inteligencia artificial permite analizar grandes cantidades de datos."
oracion2 = "La IA puede procesar grandes volúmenes de información."
oracion3 = "Los estudiantes juegan fútbol durante el descanso."

# Convertir las oraciones en embeddings
embeddings = modelo_similitud.encode([
    oracion1,
    oracion2,
    oracion3
])

# Calcular similitudes
similitud_1_2 = cosine_similarity(
    [embeddings[0]],
    [embeddings[1]]
)[0][0]

similitud_1_3 = cosine_similarity(
    [embeddings[0]],
    [embeddings[2]]
)[0][0]

print("Oración 1:")
print(oracion1)

print("\nOración 2:")
print(oracion2)

print("\nSimilitud oración 1 - oración 2:")
print(round(float(similitud_1_2), 4))

print("\nOración 3:")
print(oracion3)

print("\nSimilitud oración 1 - oración 3:")
print(round(float(similitud_1_3), 4))
```

### Análisis del resultado

Las dos primeras oraciones presentan una similitud alta debido a que expresan una idea relacionada: utilizar inteligencia artificial para procesar grandes cantidades de información.

La tercera oración habla sobre un tema diferente, por lo que obtiene un valor de similitud menor.

Este resultado demuestra que los modelos pueden comparar textos considerando su significado y no solamente las palabras que contienen.

# 15. Text Classification

Text Classification consiste en asignar automáticamente una categoría a un texto.

Algunos ejemplos de clasificación son:

- Positivo o negativo.
- Spam o no spam.
- Deportes, tecnología o economía.
- Reclamo, solicitud o felicitación.

En este ejemplo se realizará una clasificación de sentimiento.

```python
from transformers import pipeline

clasificador = pipeline(
    "sentiment-analysis",
    model="nlptown/bert-base-multilingual-uncased-sentiment"
)

textos = [
    "El proyecto funcionó perfectamente y estoy muy satisfecho con el resultado.",
    "El sistema presenta muchos errores y funciona muy mal.",
    "El producto funciona de manera normal."
]

resultados_clasificacion = []

for texto in textos:
    resultado = clasificador(texto)[0]

    resultados_clasificacion.append([
        texto,
        resultado["label"],
        resultado["score"]
    ])

df_clasificacion = pd.DataFrame(
    resultados_clasificacion,
    columns=[
        "Texto",
        "Clasificación",
        "Confianza"
    ]
)

df_clasificacion
```

### Análisis del resultado

El modelo clasificó los textos de acuerdo con el sentimiento expresado.

El primer texto presenta un sentimiento positivo debido al uso de palabras relacionadas con satisfacción y buenos resultados.

El segundo texto expresa una opinión negativa al mencionar errores y mal funcionamiento.

# 16. Translation

La traducción automática consiste en transformar un texto escrito en un idioma a otro idioma mediante modelos de procesamiento de lenguaje natural.

Actualmente los modelos neuronales permiten realizar traducciones considerando el contexto completo de las palabras.

En este ejemplo se realizará una traducción de español a inglés.

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch

# Modelo de traducción español -> inglés
nombre_modelo_traduccion = "Helsinki-NLP/opus-mt-es-en"

tokenizer_traduccion = AutoTokenizer.from_pretrained(
    nombre_modelo_traduccion
)

modelo_traduccion = AutoModelForSeq2SeqLM.from_pretrained(
    nombre_modelo_traduccion
)

texto_espanol = """
La inteligencia artificial está transformando
los procesos industriales.
"""

# Tokenizar
entradas_traduccion = tokenizer_traduccion(
    texto_espanol,
    return_tensors="pt",
    truncation=True
)

# Generar traducción
with torch.no_grad():
    salida_traduccion = modelo_traduccion.generate(
        **entradas_traduccion,
        max_new_tokens=100
    )

# Convertir tokens a texto
texto_ingles = tokenizer_traduccion.decode(
    salida_traduccion[0],
    skip_special_tokens=True
)

print("TEXTO ORIGINAL:")
print(texto_espanol)

print("\nTRADUCCIÓN AL INGLÉS:")
print(texto_ingles)
```

### Análisis del resultado

El modelo convirtió automáticamente el texto escrito en español a una representación equivalente en inglés.

La traducción neuronal busca mantener no solamente las palabras individuales sino también el significado de la oración.

Esta tecnología es utilizada en plataformas multilingües, sistemas de comunicación y procesamiento internacional de documentos.

# 17. Text Generation

Text Generation consiste en generar automáticamente una continuación a partir de un texto inicial.

El modelo analiza las palabras proporcionadas y predice cuáles pueden aparecer posteriormente.

A diferencia del ejemplo anterior con un LLM basado en instrucciones, en este apartado se estudia la generación como continuación de una secuencia de texto.

```python
from transformers import pipeline, set_seed

generador_texto = pipeline(
    "text-generation",
    model="gpt2"
)

set_seed(42)

texto_inicial = "Artificial intelligence will transform the future because"

resultado_generacion = generador_texto(
    texto_inicial,
    max_new_tokens=60,
    num_return_sequences=1,
    do_sample=True,
    temperature=0.8
)

print("Texto inicial:")
print(texto_inicial)

print("\nTexto generado:")
print(resultado_generacion[0]["generated_text"])
```

### Análisis del resultado

El modelo utilizó el texto inicial como contexto para predecir nuevas palabras y continuar la secuencia.

A diferencia de técnicas como clasificación o análisis de frecuencia, en la generación de texto el sistema produce nuevo contenido.

Debido a que se utiliza un componente probabilístico, diferentes ejecuciones pueden producir resultados distintos. Esto demuestra que los modelos generativos no recuperan únicamente una respuesta fija, sino que construyen nuevas secuencias a partir de patrones aprendidos durante su entrenamiento.
